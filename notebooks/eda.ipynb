# Summary Statistics
Examining central tendency and variability of numerical variables.

```python
df.describe()
```

```python
df.describe(include='object')
```

```python
num_cols = df.select_dtypes(include=np.number).columns

num_cols
```

```python
for col in num_cols:
    plt.figure(figsize=(6,4))
    sns.histplot(df[col], kde=True)
    plt.title(col)
    plt.show()
```

```python
cat_cols = df.select_dtypes(include='object').columns

cat_cols
```

```python
for col in cat_cols:
    plt.figure(figsize=(8,4))

    df[col].value_counts().head(10).plot(
        kind='bar'
    )

    plt.title(col)
    plt.show()
```

```python
corr = df[num_cols].corr()

corr
```

```python
plt.figure(figsize=(10,8))

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm'
)

plt.title("Correlation Matrix")
plt.show()
```

```python
missing = df.isnull().sum()

missing[missing > 0]
```

```python
missing_pct = (
    df.isnull().mean() * 100
).sort_values(ascending=False)

missing_pct
```

```python
plt.figure(figsize=(8,4))

sns.heatmap(
    df.isnull(),
    cbar=False
)

plt.show()
```

```python
for col in num_cols:
    plt.figure(figsize=(6,3))

    sns.boxplot(
        x=df[col]
    )

    plt.title(col)

    plt.show()
```

```python
df['TransactionStartTime'] = pd.to_datetime(
    df['TransactionStartTime']
)
```

```python
df['hour'] = df['TransactionStartTime'].dt.hour

df['day_of_week'] = df['TransactionStartTime'].dt.day_name()

df['month'] = df['TransactionStartTime'].dt.month
```

```python
sns.countplot(
    x='hour',
    data=df
)

plt.xticks(rotation=90)
plt.show()
```

# Key Insights

## Insight 1

Transaction amounts and values are highly right-skewed, indicating that most transactions are relatively small while a small number of transactions account for very large values.

## Insight 2

The dataset contains several categorical variables with highly dominant categories, suggesting potential imbalance and the need for encoding strategies during feature engineering.

## Insight 3

The fraud label is significantly imbalanced, with non-fraudulent transactions representing the vast majority of observations.

## Insight 4

Several numerical features contain extreme outliers that may influence model performance and should be handled appropriately during preprocessing.

## Insight 5

Transaction timestamps reveal temporal purchasing patterns that may provide predictive signals when transformed into behavioral features such as transaction hour, day of week, and recency.
